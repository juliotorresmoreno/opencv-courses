## 5. Análisis de inventarios (Pandas)


* Calcula la rotación de inventarios para cada tienda. Esto se hace dividiendo las ventas totales por el stock disponible de cada producto.
* Almacena los resultados en una nueva columna dentro del DataFrame de inventarios.
* Filtra y muestra las tiendas con niveles críticos de inventario, es decir, aquellas tiendas donde el porcentaje de productos vendidos sea menor al 10% del stock disponible.
* Utiliza groupby() y operaciones matemáticas para calcular la rotación de inventarios.
* Aplica filtros con Pandas para identificar las tiendas con niveles críticos.

```python
sales_df
```

```python
ventas_productos = sales_df[["Producto", "Cantidad_Vendida"]].groupby(by="Producto", as_index=False).sum()
ventas_productos
```

```python
stock_disponible = inventories_df[["Producto", "Stock_Disponible"]].groupby(by="Producto", as_index=False).sum()
stock_disponible
```

```python
df_temp = ventas_productos.merge(stock_disponible, on="Producto", how="inner")
df_temp["Rotacion"] = df_temp["Cantidad_Vendida"] / df_temp["Stock_Disponible"]
df_temp
```

```python
inventories_df["ID_Tienda"] = inventories_df["ID_Tienda"].astype(str)
```

```python
df_temp = sales_df[["ID_Tienda", "Producto", "Cantidad_Vendida"]].merge(
    inventories_df[["ID_Tienda", "Producto", "Stock_Disponible"]],
    on=["Producto", "ID_Tienda"],
    how="inner"
)

df_temp["Rotacion"] = df_temp["Cantidad_Vendida"] / df_temp["Stock_Disponible"]
df_temp["Critico"] = df_temp["Rotacion"] < .1
df_temp
```

```python
df_temp = sales_df[["ID_Tienda", "Producto", "Cantidad_Vendida"]].merge(
    inventories_df[["ID_Tienda", "Producto", "Stock_Disponible"]],
    on=["Producto", "ID_Tienda"],
    how="inner"
)

df_temp = df_temp.groupby("ID_Tienda", as_index=False).sum()

df_temp["Rotacion"] = df_temp["Cantidad_Vendida"] / df_temp["Stock_Disponible"]
df_temp["Critico"] = df_temp["Rotacion"] < .1

rotacion_df = df_temp

df_temp
```

## 6. Satisfacción del cliente (Pandas)


* Realiza un análisis de la satisfacción de los clientes en cada tienda. Relaciona estos datos con el rendimiento de las ventas.
* Filtra las tiendas con niveles bajos de satisfacción (< 60%) y haz recomendaciones para mejorar el rendimiento de estas tiendas.

```python
satisfaction_df["ID_Tienda"] = satisfaction_df["ID_Tienda"].astype(str)

df_temp = rotacion_df[[
    "ID_Tienda", "Cantidad_Vendida", "Stock_Disponible", "Rotacion", "Critico"
]].merge(
    satisfaction_df[["ID_Tienda", "Satisfacción_Promedio"]],
    on=["ID_Tienda"],
    how="inner"
)
df_temp[["baja_satisfaccion"]] = df_temp[["Satisfacción_Promedio"]] < 60

df_temp
```

```python
df_temp["Cantidad_Vendida"].corr(df_temp["Satisfacción_Promedio"])
```

Conclusiones

Se observa una correlación negativa pero débil entre la satisfacción promedio y la cantidad vendida. Esto sugiere que, en algunos casos, las tiendas con menor nivel de satisfacción podrían estar intentando mejorar su desempeño comercial aumentando la oferta de productos o promoviendo mayores volúmenes de venta.

Otra posible interpretación es que las tiendas con mayor flujo de clientes experimenten más congestión, tiempos de espera más largos o mayor carga operativa, lo que podría impactar negativamente la experiencia del usuario y, por ende, su nivel de satisfacción.

Adicionalmente, existen otros factores que podrían influir en esta relación y que no han sido analizados en detalle, como la ubicación de cada tienda, el perfil socioeconómico de la zona, la competencia cercana o las estrategias de precios. Estos elementos podrían ayudar a explicar mejor las diferencias observadas entre ventas y satisfacción.
